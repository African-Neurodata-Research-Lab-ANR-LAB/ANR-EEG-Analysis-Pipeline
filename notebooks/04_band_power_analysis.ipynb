# ANR EEG Band Power Analysis

Frequency bands:

- Delta: 1-4 Hz
- Theta: 4-8 Hz
- Alpha: 8-13 Hz
- Beta: 13-30 Hz
- Gamma: >30 Hz


```python
import numpy as np
import matplotlib.pyplot as plt

bands = ["Delta","Theta","Alpha","Beta","Gamma"]
values = np.zeros(len(bands))

plt.bar(bands, values)
plt.title("ANR EEG Band Power")
plt.ylabel("Power")
plt.show()
```
